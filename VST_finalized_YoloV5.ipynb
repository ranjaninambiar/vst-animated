```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2-large")
```

```python
model = GPT2LMHeadModel.from_pretrained("gpt2-large", pad_token_id=tokenizer.eos_token_id)
```

```python
tokenizer.decode(tokenizer.eos_token_id)
```

```python
input_ids=tokenizer.encode (nlp(name), return_tensors='pt')
```

```python
output=model.generate(input_ids, max_length=100, num_beams=5, no_repeat_ngram_size=2, early_stopping=True)
```

```python
tokenizer.decode (output[0], skip_special_tokens=True)
```

```python
input_ids2=tokenizer.encode (nlp3(name), return_tensors='pt')
```

```python
output1=model.generate(input_ids2, max_length=100, num_beams=5, no_repeat_ngram_size=2, early_stopping=True)
val2=tokenizer.decode (output1[0], skip_special_tokens=True)
```

```python
from nltk.translate.bleu_score import sentence_bleu
reference = [['man', 'standing',"near","the","window"]]
score = sentence_bleu(reference,val1)
print(score)
```
